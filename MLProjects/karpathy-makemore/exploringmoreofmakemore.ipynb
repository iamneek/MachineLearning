```python
words = open('names.txt', 'r', encoding="utf-8").read().splitlines()
```

```python
import re
cleaned_words = []
for w in words:
    w = w.strip().lower()
    w = re.sub(r'[^a-z]', '', w)
    if w:
        cleaned_words.append(w)
chars = sorted(list(set(''.join(cleaned_words))))
print(f"Unique characters: {len(chars)}")
```
