```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
vec_a = np.array([15, 8])
vec_b = np.array([10, 17])
```

```python
ax = plt.axes()
plt.xlim(0, 30)
plt.ylim(0, 30)
ax.arrow(0.0, 0.0, vec_a[0], vec_a[1], head_width=0.7, head_length=0.8)
plt.annotate(f"vec_a({vec_a[0]}, {vec_a[1]})", xy=(vec_a[0], vec_a[1]), xytext=(vec_a[0]+1, vec_a[1]+0.5))
ax.arrow(0.0, 0.0, vec_b[0], vec_b[1], head_width=0.7, head_length=0.8)
plt.annotate(f"vec_b({vec_b[0]}, {vec_b[1]})", xy=(vec_b[0], vec_b[1]), xytext=(vec_b[0]+1, vec_b[1]+0.5))

plt.show()
```

```python
print("Measuring cosine similarity")
sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a)*np.linalg.norm(vec_b))
print("Cosine Similarity Between vec_a and vec_b: ", sim)
print("Cosine Distance: ", 1-sim)
```
